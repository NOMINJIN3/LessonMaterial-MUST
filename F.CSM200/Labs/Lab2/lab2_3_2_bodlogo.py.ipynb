```python
# bodlogo.py
import math


class Bodlogo:

    def avg(self, a, b, c):
        return (a + b + c) / 3

    def quadratic_solver(self, a, b, c):
        if a == 0:
            return "a нь 0 байж болохгүй (квадрат тэгшитгэл биш)."

        d = b**2 - 4 * a * c
        if d > 0:
            x1 = (-b + math.sqrt(d)) / (2 * a)
            x2 = (-b - math.sqrt(d)) / (2 * a)
            return f"Шийдүүд: x1 = {x1:.2f}, x2 = {x2:.2f}"
        elif d == 0:
            x = -b / (2 * a)
            return f"Нэг шийдтэй: x = {x:.2f}"
        else:
            return "Бодит шийдгүй."
```
